```python
# 專門處理 Shioaji usage tuple 格式
usage_data = [d for d in shioaji_service.api.usage()]

print("📊 Shioaji API 使用量統計")
print("=" * 40)

# 轉換為字典方便處理
usage_dict = dict(usage_data)

# 顯示連接次數
if "connections" in usage_dict:
    print(f"🔗 連接次數: {usage_dict['connections']:,}")

# 顯示數據使用量
if "bytes" in usage_dict:
    used_mb = usage_dict["bytes"] / (1024 * 1024)
    print(f"💾 已使用流量: {used_mb:.2f} MB")

if "limit_bytes" in usage_dict:
    limit_mb = usage_dict["limit_bytes"] / (1024 * 1024)
    print(f"📊 流量限制: {limit_mb:.2f} MB")

if "remaining_bytes" in usage_dict:
    remaining_mb = usage_dict["remaining_bytes"] / (1024 * 1024)
    if remaining_mb < 0:
        print(f"⚠️  超過限制: {abs(remaining_mb):.2f} MB")
    else:
        print(f"⭐ 剩餘流量: {remaining_mb:.2f} MB")

# 計算使用率
if "bytes" in usage_dict and "limit_bytes" in usage_dict:
    usage_rate = (usage_dict["bytes"] / usage_dict["limit_bytes"]) * 100
    print(f"📈 使用率: {usage_rate:.1f}%")

    if usage_rate > 100:
        print("🚨 警告: 已超過流量限制!")
    elif usage_rate > 90:
        print("⚠️  警告: 接近流量限制!")

print("=" * 40)
```
