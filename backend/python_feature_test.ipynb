```python
class Query:
    def __init__(self, query: str):
        self.query = query

    def __enter__(self):
        return self

    def __exit__(self, exc_type, exc_value, traceback):
        if exc_type:
            print(f"Error: {exc_type} {exc_value} {traceback}")
        else:
            print("No error")

    def __str__(self):
        return self.query

```

```python
with Query("select * from users") as q:
    print(q)
```
